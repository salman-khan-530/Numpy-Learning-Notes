# What is Numpy

NumPy stands for Numerical Python.

NumPy is a Python library used for fast numerical and scientific computing.

It is mainly used for:

Mathematical calculations

Working with large datasets

Linear algebra

Machine Learning

Data Science

AI development

## Why NumPy is Used in AI/ML

NumPy is used in Artificial Intelligence and Machine Learning because these fields require a lot of mathematical calculations on large datasets.

NumPy helps us perform these calculations very fast and efficiently.

## What is numpy array

A NumPy Array is the main data structure in NumPy.

It is a grid of values (numbers) stored in a structured way so that computers can perform very fast mathematical operations.

## Numpy array vs Python list
Both Python lists and NumPy arrays can store sequences of numbers, but NumPy arrays are much better for AI/ML tasks.
Because:

Python lists are slow for math

Lists cannot handle matrix operations easily

Lists are not optimized for ML-level calculations
